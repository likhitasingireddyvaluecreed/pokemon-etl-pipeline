# Pokémon Data Analytics ETL Project

### End-to-End Data Engineering Pipeline

This project demonstrates an end-to-end **ETL (Extract, Transform, Load) pipeline** using Pokémon data from the **PokéAPI**.

The pipeline:

- **Extracts** Pokémon, species, and move data from an API
- **Stores raw data** in Parquet format
- **Transforms and cleans** the data using Python and Pandas
- **Validates** data quality and relationships
- **Loads** the processed data into a normalized SQLite database
- **Analyzes** the processed data using Pandas and visualizations

### Technologies Used

**Python · Requests · Pandas · PyArrow · SQLite · SQL · Matplotlib**

---

### Project Flow

```text
PokéAPI
   ↓
Extract
   ↓
Raw Parquet Files
   ↓
Transform + Clean
   ↓
Data Validation
   ↓
Normalized SQLite Database
   ↓
Data Analysis & Visualization
```

````python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
````

````python
from pathlib import Path

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "pokemon.db").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DB_PATH = PROJECT_ROOT / "pokemon.db"

conn = sqlite3.connect(DB_PATH)

print("Connected to SQLite database successfully!")
print("Database:", DB_PATH)
````

````python
pokemon_df = pd.read_sql_query(
    "SELECT * FROM pokemon",
    conn
)

species_df = pd.read_sql_query(
    "SELECT * FROM pokemon_species",
    conn
)

types_df = pd.read_sql_query(
    "SELECT * FROM pokemon_types",
    conn
)

abilities_df = pd.read_sql_query(
    "SELECT * FROM pokemon_abilities",
    conn
)

moves_df = pd.read_sql_query(
    "SELECT * FROM moves",
    conn
)

pokemon_moves_df = pd.read_sql_query(
    "SELECT * FROM pokemon_moves",
    conn
)

print("All tables loaded successfully!")
````

````python
table_sizes = pd.DataFrame({
    "Table": [
        "pokemon",
        "pokemon_species",
        "pokemon_types",
        "pokemon_abilities",
        "moves",
        "pokemon_moves"
    ],
    "Rows": [
        len(pokemon_df),
        len(species_df),
        len(types_df),
        len(abilities_df),
        len(moves_df),
        len(pokemon_moves_df)
    ]
})

table_sizes
````
